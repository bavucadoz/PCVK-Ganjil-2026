4. Citra yang dibaca dengan skimage.io.imread() ditampilkan berdampingan
dengan hasil cv2.cvtColor(img, cv2.COLOR_BGR2RGB) atas citra yang
sama. Manakah yang menampilkan warna sebenarnya, dan manakah yang
kanalnya tertukar? Buktikan dengan nilai piksel, bukan dengan pengamatan
visual saja.
    
    = skimage.io.imread() menampilkan warna sebenarnya karena membaca citra dalam format RGB. Sedangkan cv2.imread() membaca dalam format BGR, sehingga kanal merah dan biru tertukar jika langsung ditampilkan. Setelah menggunakan cv2.cvtColor(img, cv2.COLOR_BGR2RGB), hasil OpenCV menjadi RGB dan warnanya sama dengan skimage. Hal ini dapat dibuktikan dari nilai piksel, karena nilai piksel pada kedua gambar sama setelah konversi, berarti urutan kanalnya sudah benar.

```python
path = 'wonyoung1.jpg'

# Baca dengan OpenCV -> BGR
img_bgr = cv2.imread(path)

# Konversi BGR -> RGB
img_rgb_cv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Baca langsung menggunakan skimage -> RGB
img_rgb_skimage = io.imread(path)

# Tampilkan
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb_skimage)
plt.title("skimage.io.imread()")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_rgb_cv)
plt.title("cv2.cvtColor(BGR -> RGB)")
plt.axis("off")

plt.show()

# Periksa piksel pada koordinat (100, 100)
y, x = 100, 100

print("Koordinat:", (x, y))
print("OpenCV BGR             :", img_bgr[y, x])
print("OpenCV setelah konversi :", img_rgb_cv[y, x])
print("skimage RGB             :", img_rgb_skimage[y, x])

print("\nApakah hasil RGB sama?")
print(np.array_equal(img_rgb_cv, img_rgb_skimage))
```

Rumuskan satu pertanyaan tentang citra digital yang dapat Anda jawab
menggunakan kode dari modul ini. Contoh pertanyaan: berapa persen
piksel pada foto KTM Anda yang lebih gelap dari nilai 100? apakah
mengecilkan lalu membesarkan citra mengubah rata-rata intensitasnya?
Tuliskan pertanyaan Anda, rancangan program, data hasilnya, dan
kesimpulan Anda.

```python
img = cv2.imread('wonyoung1.jpg')

# Ukuran awal
h, w = img.shape[:2]

# Ukuran diperkecil menjadi 50%
w_kecil = int(w * 0.5)
h_kecil = int(h * 0.5)

# Resize menjadi setengah ukuran
kecil = cv2.resize(
    img,
    (w_kecil, h_kecil),
    interpolation=cv2.INTER_AREA
)

# Kembalikan ke ukuran awal
besar_kembali = cv2.resize(
    kecil,
    (w, h),
    interpolation=cv2.INTER_CUBIC
)

# Rata-rata intensitas
mean_awal = img.mean()
mean_kecil = kecil.mean()
mean_kembali = besar_kembali.mean()

print("=== UKURAN CITRA ===")
print("Ukuran awal        :", img.shape)
print("Setelah diperkecil :", kecil.shape)
print("Setelah diperbesar :", besar_kembali.shape)

print("\n=== RATA-RATA INTENSITAS ===")
print("Mean citra awal        :", mean_awal)
print("Mean citra diperkecil  :", mean_kecil)
print("Mean citra diperbesar  :", mean_kembali)

print("\n=== PERUBAHAN MEAN ===")
print(
    "Perubahan mean setelah diperkecil:",
    abs(mean_kecil - mean_awal)
)

print(
    "Perubahan mean setelah diperbesar:",
    abs(mean_kembali - mean_awal)
)
```

```python
#Visualisasi

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Citra Awal")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(kecil, cv2.COLOR_BGR2RGB))
plt.title("Setelah Diperkecil")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(besar_kembali, cv2.COLOR_BGR2RGB))
plt.title("Diperbesar Kembali")
plt.axis("off")

plt.show()
```

**Kesimpulan Analisa**

Berdasarkan hasil program, citra yang diperkecil menggunakan cv2.resize() memiliki jumlah piksel yang lebih sedikit dibandingkan citra awal. Setelah diperbesar kembali, dimensinya dapat dikembalikan ke ukuran awal, tetapi informasi detail yang telah hilang ketika proses pengecilan tidak dapat dikembalikan sepenuhnya.

Nilai rata-rata intensitasnya juga dapat mengalami perubahan setelah proses pengecilan dan pembesaran kembali. Hal tersebut terjadi karena proses resizing menggunakan interpolasi untuk memperkirakan nilai piksel baru.

Itu artinya, mengubah ukuran citra bukan hanya mengubah jumlah piksel, tetapi dapat menyebabkan perubahan pada representasi nilai piksel. Modul menjelaskan bahwa interpolasi tidak menciptakan kembali informasi yang telah hilang pada tahap sampling.

**Studi Kasus 1**

```python
def tambah_label(path, nama):
    img = cv2.imread(path)

    if img is None:
        print("Citra tidak dapat dibaca:", path)
        return None
    h, w = img.shape[:2]


    # Ukuran elemen adaptif
    # Tinggi bar = 12% tinggi gambar
    bar_h = max(1, int(0.12 * h))

    # Ketebalan = 1% ukuran terkecil
    thickness = max(1, int(0.01 * min(h, w)))

    # Ukuran font berdasarkan ukuran citra
    font_scale = max(0.5, min(h, w) / 700)


    # Membuat bar identitas
    # Salinan citra
    hasil = img.copy()

    # Bar hitam di bagian bawah
    hasil[h-bar_h:h, :] = (0, 0, 0)

    # Posisi teks
    x_text = max(1, int(0.03 * w))
    y_text = h - int(0.04 * h)

    # Teks
    cv2.putText(
        hasil,
        nama,
        (x_text, y_text),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (255, 255, 255),
        thickness,
        cv2.LINE_AA
    )
    return hasil

foto1 = tambah_label('wonyoung1.jpg', "Jang Wonyoung")
foto2 = tambah_label('fotome.jpg', "Arya Bayu Samodra")

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(foto1, cv2.COLOR_BGR2RGB))
plt.title("Foto 1 - Label Adaptif")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(foto2, cv2.COLOR_BGR2RGB))
plt.title("Foto 2 - Label Adaptif")
plt.axis("off")

plt.show()
```

**Studi Kasus 2**

```python
# Studi Kasus 2

def buat_thumbnail(nama_file):
    img = cv2.imread(nama_file)
    h, w = img.shape[:2]

    # Ukuran canvas persegi
    ukuran = max(h, w)

    # Membuat canvas putih
    canvas = np.ones((ukuran, ukuran, 3), dtype=np.uint8) * 255

    # Resize tanpa mengubah rasio
    if w > h:
        new_w = ukuran
        new_h = int(h * ukuran / w)
    else:
        new_h = ukuran
        new_w = int(w * ukuran / h)

    foto = cv2.resize(img, (new_w, new_h))

    # Posisi foto di tengah
    x = (ukuran - new_w) // 2
    y = (ukuran - new_h) // 2

    canvas[y:y+new_h, x:x+new_w] = foto

    # Nama toko
    cv2.putText(
        canvas,
        "BAE'U Official Store",
        (ukuran//20, ukuran//10),
        cv2.FONT_HERSHEY_SIMPLEX,
        ukuran/1000,
        (0,0,0),
        2
    )
    return canvas

foto1 = buat_thumbnail("ktm-biasa.jpg")
foto2 = buat_thumbnail("fotome.jpg")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(foto1, cv2.COLOR_BGR2RGB))
plt.title("Foto 1")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(foto2, cv2.COLOR_BGR2RGB))
plt.title("Foto 2")
plt.axis("off")

plt.show()
```
